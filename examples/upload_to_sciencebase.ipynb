## Parent Item

Now we are going to get the parent item, which returns a JSON file that describes the Science Base page. You can adjust this to adjust the web page.

```python
item = session.get_item(parent_id)
```

## Upload MTH5s

We are going to upload the MTH5 files to the parent page, which is different from what we have done in the past and should be easier for the user. Effectively, we're going to upload the file in question and then attach it to the parent item on ScienceBase

It helps greatly to have all of your MTH5 and XML files in one directory for both convenience and organizational elegance. Set the path to that directory. This loop uploads all the .h5 files in the directory in question; if you came from the other notebook and want to upload your XMLs as well, just switch the glob in the first line from ``*.h5`` to ``*.xml``.

<div class="alert alert-block alert-warning">
<b>Warning:</b> Sometimes there are timeouts or something goes wrong with uploading, so our for block in the next cell will catch the error and move on to the next file. Once it's done running, make sure to check the output for MTH5 files that were not uploaded properly.
</div>

```python
h5_path = Path(r"c:\\Users\\jpeacock\\OneDrive - DOI\\MTData\\CL2021\\archive")
```

```python
for h5_file in list(h5_path.glob("*.h5")):
    if h5_file.name in [d["name"] for d in item["files"]]:
        print(f"Skipping: {h5_file.name} is already uploaded")
        continue
    try:
        item = session.upload_file_to_item(item, h5_file.as_posix(), scrape_file=False)
        print(f"{time.ctime()} Uploaded: {h5_file.name}")
    except Exception as error:
        print(f"{h5_file.name} Error: {error}")
        print(f"Is Logged In: {session.is_logged_in()}")
        item = session.get_item(parent_id)
        continue
```

### Check for failed files

If the log got deleted or you didn't run the previous cell, check for files that are not on the parent page.

```python
fns = [d["name"] for d in item["files"]] 
for a in h5_path.glob("*.h5"):
    if a.name not in fns:
        print(a.name)
```

## Reorder File List

Since files are ordered in the order they are uploaded and not necessarily in alphanumeric order, we should fix that so everything looks nice.

```python
fns = [d["name"] for d in item["files"]]
print(fns)
```

```python
initial_files = []
h5_files = []
for fn in fns:
    if fn.endswith(".xml"):
        initial_files.append(fn)
    elif fn.endswith(".zip"):
        initial_files.append(fn)
    else:
        h5_files.append(fn)

ordered_fns = initial_files + sorted(h5_files)
for fn in ordered_fns:
    print(fn)
```

## Reorder file list by name

```python
new_file_list = []

for fn in ordered_fns:
    for file_item in item["files"]:
        if file_item["name"] == fn:
            new_file_list.append(file_item)
            break

item["files"] = new_file_list
for d in item["files"]:
    print(d["name"])
```

And to finalize our uploads and attach them to the SB page, all that's left to do is run ``update_item`` with our newly organized file listing.

```python
session.update_item({"id": parent_id, "files": new_file_list})
```

The last step of course is to log out, and we're done!

```python
session.logout()
```
